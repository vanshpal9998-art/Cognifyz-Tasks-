```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

data = pd.read_csv(r"/home/vansh-kumar/cognifyz Task/Dataset .csv")

# Display the first five rows
print("First Five Rows:")
print(data.head())
```

```python
# Find the Number of Rows and Columns 

print("\nDataset Shape:")
print("Rows:",data.shape[0])
print("Columns:",data.shape[1])
```

```python
# Display Column name and Data type

print("\nColumn Names:")
print(data.columns.tolist())

print("\nData Types:")
print(data.dtypes)
```

```python
# Check Missing Value

print("\nMissing Value in Each COlumns:")
print(data.isnull().sum())

print("Total Missing Value:", data.isnull().sum().sum())
```

```python
# Check Duplicate Rows

print("\nDuplicate Rows:", data.duplicated().sum())
```

```python
# Convert Aggregate rating to numeric

data["Aggregate rating"] = pd.to_numeric(data["Aggregate rating"], errors="coerce")
data["Aggregate rating"]
```

```python
# Display descriptive statistiics

print("\nNumerical Summary:")
print(data.describe())
```

```python
# Plot the rating distribution

plt.figure(figsize=(10,5))

sns.histplot(data["Aggregate rating"].dropna(),bins=20, kde=True)

plt.title("Distribution of Aggreagte Rating Restaurant Rating")
plt.xlabel("Aggregate Rating")
plt.ylabel("Number of Restaurants")
plt.tight_layout()
plt.show()
```
